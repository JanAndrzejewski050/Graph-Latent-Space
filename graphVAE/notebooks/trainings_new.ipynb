# Universal Graph VAE Training

This notebook trains the Graph VAE architectures: **Enhanced GCN-VAE**, **GCN VAE v1**, **GAT-VAE**, and **GIN-VAE**.
It reuses the highly optimized universal training script (`src/train.py`).
The outputs will be saved to the `outputs/` directory in the project root and can be directly evaluated using the `evaluate_model.ipynb` notebook.

## Setup

```python
import os, sys

# Make sure the `src` directory is importable
PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), '..', '..'))
if PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)

from src.config import TrainConfig
from src.train import run_training

# Print out the project root so we know we are operating in the right directory
print(f"Project root: {PROJECT_ROOT}")
```

## 1. Train Enhanced GCN-VAE

This trains the new 4-layer residual GCN with multi-bond prediction. The `output_dir` from the config specifies where the model and history will be saved (defaults to `outputs/`).

```python
config_enhanced = TrainConfig(
    arch="enhanced_gcn_vae",
    epochs=25,
    batch_size=256,
    use_multi_bond=True
)

print("Training Enhanced GCN-VAE...")
model_enhanced, history_enhanced = run_training(config_enhanced)
```

## 2. Train GAT-VAE (Graph Attention Network)

This trains the SOTA Graph Attention Network variant. It uses multi-head attention to learn the importance of different neighboring atoms.

```python
config_gat = TrainConfig(
    arch="gat_vae",
    epochs=25,
    batch_size=512,
    use_multi_bond=True
)

print("\nTraining GAT-VAE...")
model_gat, history_gat = run_training(config_gat)
```

## 3. Train GIN-VAE (Graph Isomorphism Network)

This trains the SOTA Graph Isomorphism Network variant, theoretically the most expressive graph representation layer.

```python
config_gin = TrainConfig(
    arch="gin_vae",
    epochs=25,
    batch_size=512,
    use_multi_bond=True
)

print("\nTraining GIN-VAE...")
model_gin, history_gin = run_training(config_gin)
```
