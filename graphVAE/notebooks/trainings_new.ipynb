# Universal Graph VAE Training

This notebook trains the two Graph VAE architectures: **Enhanced GCN-VAE** and **GCN VAE v1**.
It reuses the highly optimized universal training script (`src/train.py`).
The outputs will be saved to the `outputs/` directory in the project root and can be directly evaluated using the `evaluate_model.ipynb` notebook.

## Setup

```python
import os, sys

PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), '..', '..'))
if PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)

from src.config import TrainConfig
from src.train import run_training

print(f"Project root: {PROJECT_ROOT}")
```

## 1. Train Enhanced GCN-VAE

This trains the new 4-layer residual GCN with multi-bond prediction. The `output_dir` from the config specifies where the model and history will be saved (defaults to `outputs/`).

```python
config_enhanced = TrainConfig(
    arch="enhanced_gcn_vae",
    epochs=25,
    batch_size=512,
    use_multi_bond=True
)

print("Training Enhanced GCN-VAE...")
model_enhanced, history_enhanced = run_training(config_enhanced)
```

## 2. Train GCN VAE v1 (Original)

This trains the original 2-layer GCN architecture wrapped in the new universal framework.

```python
config_v1 = TrainConfig(
    arch="gcn_vae_v1",
    epochs=25,
    batch_size=256,
    use_multi_bond=False
)

print("\nTraining GCN VAE v1...")
model_v1, history_v1 = run_training(config_v1)
```
